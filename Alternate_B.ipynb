```python
import sys

def alternate(s):
    tracker = 0
    result = []
    for char in s:
        if char == 'a':
            result.append(char)
        else:  
            result.append(str(tracker))   
            tracker = 1 - tracker           
    return ''.join(result)

if __name__ == "__main__":
    s = sys.stdin.readline().strip()
    result = alternate(s)

    print(result)
```
