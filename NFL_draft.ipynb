## Draft Time

The NFL draft is soon upon us and it is only a few weeks until the best college players will have their dreams come true by joining a professional (American) football team. As always your friend Bubba just won’t stop talking about it.

The NFL draft is a process where NFL teams select college players, one at a time, in rounds from a pool of total players. Bubba has checked with all his sources in the different teams and already knows how each team rates each of the college players. Knowing how the draft will end a few weeks early has left Bubba with some time to solve his favourite problem. Would it be possible be to make this draft a happy draft; a draft where all the teams and all the players are happy with the result?

Bubba thinks all teams and all players will be happy with the result if there exists no player and team such that both of the following are true:

  -  P is undrafted or would prefer to go to team over the team he got drafted by

  - T can draft more players or would prefer over one of the players they have already drafted

Thankfully all the college players keep their SnapFace profiles up to date with a ranked list of their favourite NFL teams. Can you help Bubba create a happy draft?
Input

The first line of the input consists of three integers , and representing the number of NFL teams, the number of rounds in this years draft, and the number of players in the draft.
Then follows lines. Each line begins with a word, the name of the NFL team, followed by words, the name of every player in the draft, rated from best to worst according to this team.
Then follows lines. Each line begins with a word, the name of the player, followed by words, the name of every NFL team in the draft, rated from best to worst according to this player.
Output

Output the result of a happy draft if one is possible and Hello darkness my old friend! if no happy draft can be made. A happy draft should be formatted on lines. Each line begins with a word, the name of the NFL team, followed by words, the names of the players drafted by this team. Any happy draft on this format will be accepted.

## Solution:

```python
import sys, io

sample_input = """3 2 6
browns martellus tyrod tom john teddy danny
rams danny tyrod martellus john teddy tom
giants teddy danny tyrod john tom martellus
teddy giants browns rams
danny giants browns rams
tyrod giants rams browns
john giants rams browns
tom browns giants rams
martellus giants browns rams
"""


data = sys.stdin.read().split()

```

### Test

```python
print(len(data))
print(data[:5])
```

### Map first line to N, M, K

```python
N, M, K = int(data[0]), int(data[1]), int(data[2])
pos = 3 #so we start reading from the 4th element in the list
```

```python

print("N:", N , "M:", M, "K:", K)
```

### Map team names and rankings

```python
team_names = []
team_rankings = []

for i in range(N):

    team_names.append(data[pos])

    ranking = data[pos+1:pos+K+1]
    d = {}
    for index, name in enumerate(ranking):
        d[name] = index 
    team_rankings.append(d)
    
    pos = pos + (K+1) 
        
```

```python

print(team_names)
print(team_rankings)
print(pos, data[pos])
```

### Map player preferences

```python
player_names = []
player_preferences = []

team_id = {name: index for index, name in enumerate(team_names)}


for i in range(K):

    player_names.append(data[pos])

    player_preferences.append(data[pos+1: pos + 1 + N])

    pos = pos + (N+1)
```

```python
print(team_id)              # {'browns': 0, 'rams': 1, 'giants': 2}
print(team_id['giants'])    
print(player_preferences)
```

### The algo


```python


max_team_capacity = M 

rosters = [[] for _ in range(N)]

temp_team = [None] * K

next_preference = [0] * K

free_players = list(range(K))

```

```python
print(player_preferences)
```

```python
print(free_players)
print(rosters)
```

```python
while free_players:
    p = free_players.pop()
    if next_preference[p] >= N:
        continue
    choice = player_preferences[p][next_preference[p]]   

    t = team_id[choice]                                   
    next_preference[p] += 1                             
    
    if len(rosters[t]) < max_team_capacity:
        rosters[t].append(p)
        temp_team[p] = t
    else:
        worst = max(rosters[t], key=lambda x: team_rankings[t][player_names[x]])
        if team_rankings[t][player_names[p]] < team_rankings[t][player_names[worst]]:
            rosters[t].remove(worst)
            rosters[t].append(p)
            temp_team[p] = t
            temp_team[worst] = None
            free_players.append(worst)
        else:
            free_players.append(p)

```

```python
print(player_preferences)
print(team_id)
print(rosters)
```

```python
lines = []
for t in range(N):
    highest_ranked_first = sorted(rosters[t], key=lambda x: team_rankings[t][player_names[x]])
    names = [player_names[q] for q in rosters[t]]
    lines.append(team_names[t] + " " + " ".join(names))
print("\n".join(lines))
```
