```python
def check_clause(line, n):
    literals = []
    for literal in line.strip().split(" v "):
        # Check if the literal is negated
        if literal.startswith("~"):
            is_negated = True
            literal = literal[1:]
        else:
            is_negated = False
        index = int(literal[1:]) - 1  # X1 becomes 0, X2 becomes 1, etc.
        literals.append((index, is_negated))
    return literals

#Row 1: X1 v X2 
# X1 --> 0 (and therefore false) and X2 --> 1 (and therefore true)
# Both literals are true, so the clause is satisfied.
# Outputs: Literals = [(0, False), (1, False)]

```

```python
#If bit is 1 (True) and is_negated is False, then the literal evaluates to True.
#If bit is 0 (False) and is_negated is True, then the literal also evaluates to True.

def literal_true(bit, is_negated):
    if is_negated:
        return bit == 0
    else:
        return bit == 1
```

```python
def clause_satisfied(mask, clause):
    for (index, is_negated) in clause:
        bit = (mask >> index) & 1
        if literal_true(bit, is_negated):
            return True     #If any literal is true, the clause is satisfied
    return False #If none of the literals are true, the clause is not satisfied
```

```python
#Check if a given mask satisfies all clauses
def satisfies_all(mask, clauses):
    for clause in clauses:
        if not clause_satisfied(mask, clause):
            return False
    return True

```

```python
def solve(n, clauses):
    for mask in range(1 << n):
        if satisfies_all(mask, clauses):
            return "satisfiable"
    return "unsatisfiable"
```

```python
def main():
    lines = int(input())
    results = []
    for line in range(lines):
        n, m = map(int, input().split())
        clauses = []
        for line in range(m):
            clauses.append(check_clause(input(), n))
        results.append(solve(n, clauses))
    print("\n".join(results))

main()
```
