# Greedy: Minimum Scalar Product

### problem

The input gives two test cases, given by the first line, followed by the dimension of the set of vectors and then the vectors.

Going line by line:

1.      2       (Two test cases)
2.      3       (Shape of Test1)
3.      -1 3 -5 (Vector 1)
4.      -2 4 2  (Vector 2)
5.      5       (Shape of Test2)
6.      1 2 3 4 5 (Vector 1)
7.      1 0 1 01 (Vector 2)

In the case of test1 and vectors: $v_1$ =  (1,3,5) , $v_2$ = (-2 4 1) , the minimum scalar product is -25. To make the smallest possible number, we have to pair the highest negative number with the highest positive number, as two negatives lead to a positive result, and two positives also lead to a positive result. 

Therefore the strategy is to go through both vectors, sort one by ascending and the other by descending, by which we always multiply the extremes - the highest positive or least negative will be multiplied with the least positive or highest negative.

```python
data = """
2
3
1 3 -5
-2 4 1
5
1 2 3 4 5
1 0 1 0 1"""
```

### read-in

```python
import sys, io

# toggle 

# lines = sys.stdin.read().splitlines()   # real Kattis submission
lines = data.strip().splitlines()         # local testing

print(lines)
```

### algo

```python
cursor = 0
tracker = int(lines[cursor]); cursor += 1

test_cases = []

for line in range(tracker):
    n = int(lines[cursor])
    cursor += 1
    a = list(map(int, lines[cursor].split()))
    cursor += 1
    b = list(map(int, lines[cursor].split()))
    cursor += 1
    test_cases.append((n, a, b))


#print(cursor, test_cases)
```

```python
results = []
for n, a, b in test_cases:
    a.sort()
    b.sort(reverse=True)
    scalar_product = sum(x * y for x, y in zip(a, b))
    results.append(scalar_product)

```

### output

```python
for i in range(len(results)):
    print(f"Case #{i+1}: {results[i]}")
```
