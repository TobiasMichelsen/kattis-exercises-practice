```python
# Input

import sys

n, t = map(int, sys.stdin.readline().split())
A = list(map(int, sys.stdin.readline().split()))
```

```python
def action_sum_pair(A):
    # t = 1: "Yes" if two distinct elements x != y sum to 7777, else "No"
    seen = set(A)
    for i in A:
        if 7777 - i in seen and 7777 - i != i:
            return "Yes"
    return "No"
        
```

```python
assert action_sum_pair([1, 7770, 3, 4, 5, 6, 7]) == "Yes"    # Sample Input 1
assert action_sum_pair([1, 2, 3, 4, 5, 6, 7]) == "No"         # no pair sums to 7777
assert action_sum_pair([3888, 3889, 1, 2, 3, 4, 5]) == "Yes"  # exact pair: 3888 + 3889 = 7777
assert action_sum_pair([100, 200, 300]) == "No"               # nowhere close

print("All action_sum_pair tests passed!")
```

```python
def action_uniqueness(A):
    # t = 2: "Unique" if all elements are distinct, else "Contains duplicate"
    seen = set()
    for i in A:
        if i in seen:
            return "Contains duplicate"
        seen.add(i)
    return "Unique"
```

```python
assert action_uniqueness([1, 7770, 3, 4, 5, 6, 7]) == "Unique"    
assert action_uniqueness([1,1,1,1,5,5]) == "Contains duplicate"  # duplicates present
assert action_uniqueness([100, 200, 300]) == "Unique"               

print("All action_uniqueness tests passed!")
```
